### Using Pandas

Pandas is a powerful open-source data analysis and manipulation library for Python. It provides data structures like Series and DataFrames that are designed for working with tabular and time-series data.

#### 1. Introduction to Pandas Data Structures

Pandas primarily offers two data structures: `Series` and `DataFrame`.

*   **Series**: A one-dimensional labeled array capable of holding any data type.
*   **DataFrame**: A two-dimensional labeled data structure with columns of potentially different types. You can think of it like a spreadsheet or SQL table.

```python
import pandas as pd
import numpy as np

print("--- Creating a Series ---")
s = pd.Series([1, 3, 5, np.nan, 6, 8], index=['a', 'b', 'c', 'd', 'e', 'f'])
display(s)

print("\n--- Creating a DataFrame from a dictionary ---")
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [24, 27, 22, 32, 29],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix'],
    'Occupation': ['Engineer', 'Doctor', 'Artist', 'Scientist', 'Manager'],
    'Salary': [70000, 90000, 50000, 85000, 95000]
}
df = pd.DataFrame(data)
display(df)

print("\n--- Creating a DataFrame with a date range index ---")
dates = pd.date_range('20230101', periods=6)
df_ts = pd.DataFrame(np.random.randn(6, 4), index=dates, columns=list('ABCD'))
display(df_ts)
```

#### 2. Data Loading and Saving

Pandas supports reading and writing data from various formats like CSV, Excel, SQL databases, JSON, etc.

```python
# Create a dummy CSV file for demonstration
dummy_data = {
    'StudentID': [1, 2, 3, 4, 5],
    'Subject': ['Math', 'Science', 'Math', 'Art', 'Science'],
    'Score': [85, 92, 78, 65, 88]
}
dummy_df = pd.DataFrame(dummy_data)
dummy_df.to_csv('students.csv', index=False)

print("--- Loading data from a CSV file ---")
df_from_csv = pd.read_csv('students.csv')
display(df_from_csv)

# You can also save DataFrames to various formats:
# df.to_excel('output.xlsx', index=False)
# df.to_json('output.json', orient='records')
```

#### 3. Data Inspection and Basic Information

Understanding your data's structure, types, and summary statistics is a crucial first step.

```python
print("--- Displaying the first 5 rows (head) ---")
display(df.head())

print("\n--- Displaying the last 5 rows (tail) ---")
display(df.tail())

print("\n--- Getting a concise summary of the DataFrame (info) ---")
df.info()

print("\n--- Generating descriptive statistics (describe) ---")
display(df.describe())

print("\n--- Checking the DataFrame's shape (rows, columns) ---")
print(df.shape)

print("\n--- Listing column names ---")
print(df.columns)
```

#### 4. Data Selection and Indexing

Accessing specific rows, columns, or cells in a DataFrame.

```python
print("--- Selecting a single column ---")
display(df['Name'])

print("\n--- Selecting multiple columns ---")
display(df[['Name', 'City', 'Salary']])

print("\n--- Selecting rows by label using `.loc[]` ---")
display(df.loc[0]) # Select first row
display(df.loc[1:3, ['Name', 'Age']]) # Select rows 1 to 3 and specific columns

print("\n--- Selecting rows by integer position using `.iloc[]` ---")
display(df.iloc[0]) # Select first row
display(df.iloc[1:4, 0:3]) # Select rows 1 to 3 and columns 0 to 2

print("\n--- Boolean indexing (filtering) ---")
display(df[df['Age'] > 25]) # Filter rows where Age is greater than 25
display(df[(df['City'] == 'New York') | (df['City'] == 'Chicago')]) # Multiple conditions
```

#### 5. Data Cleaning and Manipulation

Handling missing values, renaming columns, creating new columns, and dropping data.

```python
# Create a DataFrame with missing values for demonstration
df_manipulate = pd.DataFrame({
    'ID': [1, 2, 3, 4, 5],
    'Value1': [10, 20, np.nan, 40, 50],
    'Value2': [100, np.nan, 300, 400, 500],
    'Category': ['A', 'B', 'A', 'C', 'B']
})
df_manipulate.loc[0, 'Category'] = np.nan # Add another NaN
display(df_manipulate)

print("--- Checking for missing values ---")
display(df_manipulate.isnull().sum())

print("\n--- Filling missing values with a specific value ---")
df_filled = df_manipulate.fillna(0)
display(df_filled)

print("\n--- Filling missing values with the mean of the column ---")
df_filled_mean = df_manipulate['Value2'].fillna(df_manipulate['Value2'].mean())
display(df_filled_mean.to_frame()) # Convert Series back to DataFrame for display

print("\n--- Dropping rows with any missing values ---")
df_dropna = df_manipulate.dropna()
display(df_dropna)

print("\n--- Renaming columns ---")
df_renamed = df_manipulate.rename(columns={'Value1': 'Feature1', 'Value2': 'Feature2'})
display(df_renamed)

print("\n--- Creating a new column ---")
df_manipulate['New_Column'] = df_manipulate['Value1'] * 2
display(df_manipulate)

print("\n--- Applying a function to a column ---")
df_manipulate['Category_Upper'] = df_manipulate['Category'].str.upper().fillna('UNKNOWN')
display(df_manipulate)

print("\n--- Dropping columns ---")
df_dropped_col = df_manipulate.drop(columns=['New_Column', 'Category_Upper'])
display(df_dropped_col)
```

#### 6. Grouping and Aggregation

The `groupby()` method is essential for performing aggregate operations on subsets of data.

```python
print("--- Original DataFrame for Grouping ---")
display(df)

print("\n--- Group by 'City' and calculate average 'Salary' ---")
display(df.groupby('City')['Salary'].mean().to_frame())

print("\n--- Group by 'Occupation' and count number of people ---")
display(df.groupby('Occupation').size().to_frame(name='Count'))

print("\n--- Group by multiple columns and apply multiple aggregations ---")
display(df.groupby(['City', 'Occupation']).agg(
    Avg_Salary=('Salary', 'mean'),
    Min_Age=('Age', 'min'),
    Count=('Name', 'count')
))
```

#### 7. Merging, Joining, and Concatenating DataFrames

Combining DataFrames is a common task in data preparation.

```python
# Create two DataFrames for merging
df1 = pd.DataFrame({'Key': ['K0', 'K1', 'K2', 'K3'],
                    'A': ['A0', 'A1', 'A2', 'A3'],
                    'B': ['B0', 'B1', 'B2', 'B3']})

df2 = pd.DataFrame({'Key': ['K0', 'K1', 'K4', 'K5'],
                    'C': ['C0', 'C1', 'C4', 'C5'],
                    'D': ['D0', 'D1', 'D4', 'D5']})

df3 = pd.DataFrame({'A': ['A6', 'A7'],
                    'B': ['B6', 'B7']})

print("--- df1 ---")
display(df1)
print("--- df2 ---")
display(df2)
print("--- df3 ---")
display(df3)

print("\n--- Merging DataFrames (inner join on 'Key') ---")
merged_inner = pd.merge(df1, df2, on='Key', how='inner')
display(merged_inner)

print("\n--- Merging DataFrames (left join on 'Key') ---")
merged_left = pd.merge(df1, df2, on='Key', how='left')
display(merged_left)

print("\n--- Concatenating DataFrames (row-wise) ---")
concatenated_rows = pd.concat([df1[['A', 'B']], df3], ignore_index=True)
display(concatenated_rows)
```

#### 8. Time Series Functionality

Pandas has robust features for working with time-series data.

```python
# Create a time series DataFrame
idx = pd.date_range('2023-01-01', periods=100, freq='D')
ts_data = pd.Series(np.random.randn(100), index=idx)
ts_df = pd.DataFrame({'Value': ts_data})
display(ts_df.head())

print("--- Resampling (e.g., weekly mean) ---")
display(ts_df.resample('W').mean().head())

print("\n--- Shifting data ---")
display(ts_df['Value'].shift(1).head().to_frame())

print("\n--- Rolling window calculations (e.g., 3-day rolling mean) ---")
display(ts_df['Value'].rolling(window=3).mean().head().to_frame())
```
